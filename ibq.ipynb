```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.title("My Line Plot")
plt.xlabel(" X Values")
plt.ylabel(" Y Values")
plt.plot(x,y)
plt.show()
```

```python
import matplotlib.pyplot as plt

languages = ["Python", "Java", "C", "C++"]
students = [40, 25, 30, 20]

plt.bar(languages,students)
plt.title("MY BAR GRAPH PLOTTING")
plt.xlabel('languages')
plt.ylabel('students')
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [10, 15, 12, 20, 18, 25]
plt.title("MY FIRST SCATTER PLOT GRAPH")
plt.xlabel(x)
plt.ylabel(y)
plt.xticks([1,2,3,4,5,6])
plt.yticks([10,15,12,20,18,25])
plt.scatter(x,y,alpha=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt
labels = ["Python", "Java", "C", "C++"]
values = [40, 25, 20, 15]
plt.pie(values,labels=labels,autopct="%1.1f%%")
plt.title("PIE CHART DISPLAY")
plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 62, 65, 70, 72, 75, 80, 85, 90, 95]
plt.hist(marks,bins=5)
plt.title("HISTOGRAM DISPLAY")
plt.xlabel('marks')
plt.ylabel("frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt

languages = ["Python", "Java", "C", "C++"]
students = [40, 25, 30, 20]

plt.barh(languages,students)
plt.title("MY BAR GRAPH PLOTTING")
plt.xlabel('languages')
plt.ylabel('students')
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]
plt.subplot(1,2,1)
plt.plot(x,y)
plt.subplot(1,2,2)
plt.bar(x,y)
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]
plt.plot(x,y,color='blue',linewidth=2,marker='o',markersize=0.5,linestyle='--',alpha=1.0)
plt.title("LINE PLOT DISPLAY")
plt.xlabel("x values")
plt.ylabel("y labels")
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]

y1 = [10, 20, 15, 30, 25]
y2 = [5, 15, 20, 25, 30]

plt.plot(x, y1, label="Line 1")
plt.plot(x, y2, label="Line 2")
plt.legend()
plt.xlabel('x value')
plt.ylabel('y value')
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x,y)
plt.title("GRID FUNCTION METHOD")
plt.xlabel('x values')
plt.ylabel('y values')
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]
plt.plot(x,y)
plt.xticks([1,2,3,4,5])
plt.yticks([10,20,15,50,25])
plt.xlabel(" x values")
plt.ylabel("y values")
plt.show()
```
